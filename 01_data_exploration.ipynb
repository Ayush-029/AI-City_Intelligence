```python
# Make sure date_time is datetime
df["date_time"] = pd.to_datetime(df["date_time"])

# Extract time-related features
df["hour"] = df["date_time"].dt.hour
df["day_of_week"] = df["date_time"].dt.dayofweek
df["month"] = df["date_time"].dt.month
df["year"] = df["date_time"].dt.year

# Weekend
df["is_weekend"] = (df["day_of_week"] >= 5).astype(int)

df.head()
```

```python
df["is_rush_hour"] = df["hour"].isin([
    7, 8, 9, 16, 17, 18
]).astype(int)

df[["hour", "is_rush_hour"]].head(20)
```

```python
def get_time_period(hour):
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"

df["time_period"] = df["hour"].apply(get_time_period)

df[["hour", "time_period"]].head(20)
```

```python
# Sort data by date 
df = df.sort_values("date_time")

# training: 2012-2017
# testing : 2018

train = df[df["year"] < 2018]
test = df[df["year"] >= 2018]

print("training data:", train.shape)
print("testing data:", test.shape)
```

```python
features = [
    "temp",
    "rain_1h",
    "snow_1h",
    "clouds_all",
    "hour",
    "day_of_week",
    "month",
    "is_weekend",
    "is_rush_hour",
    "time_period"
]

target = "traffic_volume"

x_train = train[features]
y_train = train[target]

x_test = test[features]
y_test = test[target]

print("x_train:", x_train.shape)
print("y_train:", y_train.shape)
print("x_test:", x_test.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.preprocessing import LabelEncoder
import numpy as np
import pandas as pd

# Encode categorical variables
x_train_encoded = x_train.copy()
x_test_encoded = x_test.copy()

# Encode 'time_period' if it exists
if 'time_period' in x_train_encoded.columns:
    le = LabelEncoder()
    x_train_encoded['time_period'] = le.fit_transform(x_train_encoded['time_period'].astype(str))
    x_test_encoded['time_period'] = le.transform(x_test_encoded['time_period'].astype(str))

model = LinearRegression()

model.fit(x_train_encoded, y_train)

y_pred = model.predict(x_test_encoded)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Linear Regression Performance:")
print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R²:", round(r2, 2))
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100 , 
    random_state=42,
    max_depth=20,
    n_jobs=-1
    )

rf_model.fit(x_train_encoded, y_train)

rf_pred = rf_model.predict(x_test_encoded)

rf_mae = mean_absolute_error(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_r2 = r2_score(y_test, rf_pred)

print("Random Forest Performance:")
print("MAE:", round(rf_mae, 2))
print("RMSE:", round(rf_rmse, 2))
print("R²:", round(rf_r2, 4))
```

```python
from xgboost import XGBRegressor

xgb_model = XGBRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=6,
    random_state=42,
    n_jobs=-1,
    subsample=0.8,
    colsample_bytree=0.8
)

xgb_model.fit(x_train_encoded, y_train)

xgb_pred = xgb_model.predict(x_test_encoded) 
xgb_mae = mean_absolute_error(y_test, xgb_pred)
xgb_rmse = np.sqrt(mean_squared_error(y_test, xgb_pred))
xgb_r2 = r2_score(y_test, xgb_pred)

print("XGBoost Performance:")
print("MAE:", round(xgb_mae, 2))
print("RMSE:", round(xgb_rmse, 2))
print("R²:", round(xgb_r2, 4))
```

```python
import joblib
import os

# Create models directory if it doesn't exist
os.makedirs("models", exist_ok=True)

joblib.dump(xgb_model, "models/traffic_xgboost_model.joblib")

print("XGBoost model saved successfully!")
```

```python
sample = x_test_encoded.iloc[[0]]

actual_value = y_test.iloc[0]

predicted_value = xgb_model.predict(sample)[0]

print("Traffic prediction test:")
print("Actual traffic volume:", round(actual_value))
print("Predicted traffic volume:", round(predicted_value))
print("difference:", round(abs(actual_value - predicted_value)))
```

```python
def get_congestion_level(traffic_volume):
    if traffic_volume < 2000:
        return "Low"
    elif 2000 <= traffic_volume < 4000:
        return "Moderate"
    else:
        return "High"

congestion = get_congestion_level(predicted_value)
print("predicted traffic volume;" , round(predicted_value))
print("congestion level:" , congestion)
```

```python
def calculate_risk_score(traffic_volume):
    if traffic_volume < 2000:
        return round((traffic_volume / 2000) * 30, 2)
    elif traffic_volume < 4000:
        return round(30 + ((traffic_volume - 2000) / 2000) * 40, 2)
    else:
        return round(min(100, 70 + ((traffic_volume - 4000) / 2000) * 30), 2)

risk_score = calculate_risk_score(predicted_value)

print("predicted traffic volume:", round(predicted_value))
print("calculated risk score:", risk_score)
print("risk_score:", risk_score, "/100")
```

```python
def traffic_intelligence(predicted_traffic):
    congestion_level = get_congestion_level(predicted_traffic)
    risk_score = calculate_risk_score(predicted_traffic)
    
    return {
        "predicted_traffic_volume": round(predicted_traffic),
        "congestion_level": congestion_level,
        "risk_score": risk_score
    }

result = traffic_intelligence(predicted_value)
print(result)
```
